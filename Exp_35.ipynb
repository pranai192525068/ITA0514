```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.ones((400, 600, 3), dtype=np.uint8) * 255

text = input("Enter text: ")

cv2.putText(img, text, (50, 200),
            cv2.FONT_HERSHEY_SIMPLEX, 1,
            (255, 0, 0), 2)

plt.imshow(img)
plt.title("Text on Image")
plt.axis("off")
plt.show()
```
