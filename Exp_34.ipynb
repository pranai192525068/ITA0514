```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.ones((400, 400, 3), dtype=np.uint8) * 255

cv2.circle(img, (200, 200), 100, (255, 0, 0), 3)

plt.imshow(img)
plt.title("Circle")
plt.axis("off")
plt.show()
```
