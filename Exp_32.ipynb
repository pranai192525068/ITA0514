```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.ones((500, 500, 3), dtype=np.uint8) * 255

img[20:120, 20:120] = (0, 0, 0)
img[20:120, 380:480] = (255, 0, 0)
img[380:480, 20:120] = (0, 255, 0)
img[380:480, 380:480] = (0, 0, 255)

plt.imshow(img)
plt.title("Four Colored Boxes")
plt.axis("off")
plt.show()
```
