```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.ones((400, 600, 3), dtype=np.uint8) * 255

cv2.rectangle(img, (50, 100), (550, 300), (255, 0, 0), 3)

plt.imshow(img)
plt.title("Rectangle")
plt.axis("off")
plt.show()
```
