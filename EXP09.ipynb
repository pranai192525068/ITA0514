```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()
img = cv2.imread(next(iter(uploaded)))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

big = cv2.resize(img, None, fx=2, fy=2)
small = cv2.resize(img, None, fx=0.5, fy=0.5)

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(big)
plt.title("Bigger")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(small)
plt.title("Smaller")
plt.axis("off")

plt.show()
```
