```python
import cv2
from google.colab import files
from IPython.display import Video, display

uploaded = files.upload()
filename = next(iter(uploaded))

cap = cv2.VideoCapture(filename)
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

slow = cv2.VideoWriter("slow.mp4", cv2.VideoWriter_fourcc(*"mp4v"), fps/2, (width, height))
fast = cv2.VideoWriter("fast.mp4", cv2.VideoWriter_fourcc(*"mp4v"), fps*2, (width, height))

frames = []
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frames.append(frame)

    slow.write(frame)

    if len(frames) % 2 == 0:
        fast.write(frame)

cap.release()
slow.release()
fast.release()

print("Slow Motion Video:")
display(Video("slow.mp4", embed=True))

print("Fast Motion Video:")
display(Video("fast.mp4", embed=True))
```
